```python
from pathlib import Path
import pandas as pd
import numpy as np
from rdkit import Chem, rdBase

PROJECT_ROOT = Path(
    r"E:\CODE\Python\ALL Project\Machine Learning\BeeSafe AI"
)

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "dataset_final.csv"

print("Project folder exists:", PROJECT_ROOT.exists())
print("Dataset exists:", DATA_PATH.exists())
print("RDKit version:", rdBase.rdkitVersion)
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python

# BeeSafe AI — Data Quality Assessment

print("=" * 50)
print("DATA QUALITY REPORT")
print("=" * 50)

# 1. Dataset dimensions
print("\n1. DATASET SHAPE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 2. Missing values
print("\n2. MISSING VALUES")
display(
    df.isnull().sum()
      .to_frame("Missing Values")
      .sort_values("Missing Values", ascending=False)
)

# 3. Exact duplicate rows
print("\n3. EXACT DUPLICATE ROWS")
print("Duplicates:", df.duplicated().sum())

# 4. Duplicate chemical identifiers
print("\n4. DUPLICATE PUBCHEM CIDs")
print("Repeated CID records:", df["CID"].duplicated().sum())

# 5. Binary target distribution
print("\n5. TOXICITY LABEL DISTRIBUTION")
display(
    df["label"]
      .value_counts(dropna=False)
      .sort_index()
      .to_frame("Number of Records")
)

# 6. Three-level toxicity distribution
print("\n6. PPDB LEVEL DISTRIBUTION")
display(
    df["ppdb_level"]
      .value_counts(dropna=False)
      .sort_index()
      .to_frame("Number of Records")
)

# 7. Data types
print("\n7. COLUMN DATA TYPES")
display(df.dtypes.to_frame("Data Type"))
```
